```python
import pandas as pd
import plotly.express as px

# Load the dataset if it is not already loaded:
# df = pd.read_csv("TriHemo-MCV (1).csv")

class_order = [
    "Anemic",
    "Compensated Microcytosis",
    "Non-Anemic"
]

measure_order = [
    "HGB",
    "MCV",
    "RBC",
    "HCT",
    "MCH",
    "RDW_CV"
]

measure_labels = {
    "HGB": "HGB (g/dL)",
    "MCV": "MCV (fL)",
    "RBC": "RBC (million/µL)",
    "HCT": "HCT (%)",
    "MCH": "MCH (pg)",
    "RDW_CV": "RDW-CV (%)"
}

# Select and clean the required columns
profile_df = df[["Class"] + measure_order].copy()

for column in measure_order:
    profile_df[column] = (
        profile_df[column]
        .astype(str)
        .str.strip()
        .str.replace(",", ".", regex=False)
        .pipe(pd.to_numeric, errors="coerce")
    )

# Convert the data from wide to long format
long_df = profile_df.melt(
    id_vars="Class",
    value_vars=measure_order,
    var_name="Measure",
    value_name="Value"
)

long_df = long_df.dropna(subset=["Class", "Measure", "Value"])

# Calculate each class median
class_medians = (
    long_df
    .groupby(["Class", "Measure"], as_index=False, observed=True)["Value"]
    .median()
    .rename(columns={"Value": "Median"})
)

# Calculate the overall median for each measure
overall_medians = (
    long_df
    .groupby("Measure", as_index=False)["Value"]
    .median()
    .rename(columns={"Value": "Overall_median"})
)

# Combine results and calculate an index relative to the overall median
plot_df = class_medians.merge(
    overall_medians,
    on="Measure",
    how="left"
)

plot_df["Index_vs_overall"] = (
    100 * plot_df["Median"] / plot_df["Overall_median"]
)

plot_df["Measure_label"] = plot_df["Measure"].map(measure_labels)

# Explicitly preserve the intended order
plot_df["Class"] = pd.Categorical(
    plot_df["Class"],
    categories=class_order,
    ordered=True
)

plot_df["Measure"] = pd.Categorical(
    plot_df["Measure"],
    categories=measure_order,
    ordered=True
)

plot_df = plot_df.sort_values(["Measure", "Class"])

# Create six small-multiple panels
fig = px.line(
    plot_df,
    x="Class",
    y="Median",
    facet_col="Measure_label",
    facet_col_wrap=3,
    markers=True,
    category_orders={
        "Class": class_order,
        "Measure_label": [
            measure_labels[m] for m in measure_order
        ]
    },
    hover_data={
        "Class": True,
        "Measure_label": False,
        "Median": ":.2f",
        "Overall_median": ":.2f",
        "Index_vs_overall": ":.1f"
    },
    labels={
        "Class": "Patient class",
        "Median": "Class median",
        "Overall_median": "Overall median",
        "Index_vs_overall": "Index vs overall median"
    },
    title="Red-cell profiles differ distinctly across patient classes",
    template="plotly_white",
    height=750,
    width=1250
)

# Give each panel an independent y-axis because the measures use different units
fig.update_yaxes(matches=None, rangemode="tozero")

# Improve marker and line appearance
fig.update_traces(
    marker={"size": 10},
    line={"width": 2}
)

# Remove Plotly's default "Measure_label=" prefix from panel headings
fig.for_each_annotation(
    lambda annotation: annotation.update(
        text=annotation.text.replace("Measure_label=", "")
    )
)

fig.update_xaxes(
    title_text="",
    tickangle=-20
)

fig.update_yaxes(
    title_text="Median"
)

fig.update_layout(
    title={
        "x": 0.5,
        "xanchor": "center"
    },
    margin={
        "t": 100,
        "b": 100,
        "l": 80,
        "r": 40
    },
    showlegend=False
)

fig.show()

# Optional interactive export
# fig.write_html("red_cell_profiles_by_class.html")

# Optional static export; requires: pip install kaleido
# fig.write_image(
#     "red_cell_profiles_by_class.png",
#     width=1250,
#     height=750,
#     scale=2
# )import pandas as pd
import plotly.express as px

# Load the dataset if it is not already loaded:
# df = pd.read_csv("TriHemo-MCV (1).csv")

class_order = [
    "Anemic",
    "Compensated Microcytosis",
    "Non-Anemic"
]

measure_order = [
    "HGB",
    "MCV",
    "RBC",
    "HCT",
    "MCH",
    "RDW_CV"
]

measure_labels = {
    "HGB": "HGB (g/dL)",
    "MCV": "MCV (fL)",
    "RBC": "RBC (million/µL)",
    "HCT": "HCT (%)",
    "MCH": "MCH (pg)",
    "RDW_CV": "RDW-CV (%)"
}

# Select and clean the required columns
profile_df = df[["Class"] + measure_order].copy()

for column in measure_order:
    profile_df[column] = (
        profile_df[column]
        .astype(str)
        .str.strip()
        .str.replace(",", ".", regex=False)
        .pipe(pd.to_numeric, errors="coerce")
    )

# Convert the data from wide to long format
long_df = profile_df.melt(
    id_vars="Class",
    value_vars=measure_order,
    var_name="Measure",
    value_name="Value"
)

long_df = long_df.dropna(subset=["Class", "Measure", "Value"])

# Calculate each class median
class_medians = (
    long_df
    .groupby(["Class", "Measure"], as_index=False, observed=True)["Value"]
    .median()
    .rename(columns={"Value": "Median"})
)

# Calculate the overall median for each measure
overall_medians = (
    long_df
    .groupby("Measure", as_index=False)["Value"]
    .median()
    .rename(columns={"Value": "Overall_median"})
)

# Combine results and calculate an index relative to the overall median
plot_df = class_medians.merge(
    overall_medians,
    on="Measure",
    how="left"
)

plot_df["Index_vs_overall"] = (
    100 * plot_df["Median"] / plot_df["Overall_median"]
)

plot_df["Measure_label"] = plot_df["Measure"].map(measure_labels)

# Explicitly preserve the intended order
plot_df["Class"] = pd.Categorical(
    plot_df["Class"],
    categories=class_order,
    ordered=True
)

plot_df["Measure"] = pd.Categorical(
    plot_df["Measure"],
    categories=measure_order,
    ordered=True
)

plot_df = plot_df.sort_values(["Measure", "Class"])

# Create six small-multiple panels
fig = px.line(
    plot_df,
    x="Class",
    y="Median",
    facet_col="Measure_label",
    facet_col_wrap=3,
    markers=True,
    category_orders={
        "Class": class_order,
        "Measure_label": [
            measure_labels[m] for m in measure_order
        ]
    },
    hover_data={
        "Class": True,
        "Measure_label": False,
        "Median": ":.2f",
        "Overall_median": ":.2f",
        "Index_vs_overall": ":.1f"
    },
    labels={
        "Class": "Patient class",
        "Median": "Class median",
        "Overall_median": "Overall median",
        "Index_vs_overall": "Index vs overall median"
    },
    title="Red-cell profiles differ distinctly across patient classes",
    template="plotly_white",
    height=750,
    width=1250
)

# Give each panel an independent y-axis because the measures use different units
fig.update_yaxes(matches=None, rangemode="tozero")

# Improve marker and line appearance
fig.update_traces(
    marker={"size": 10},
    line={"width": 2}
)

# Remove Plotly's default "Measure_label=" prefix from panel headings
fig.for_each_annotation(
    lambda annotation: annotation.update(
        text=annotation.text.replace("Measure_label=", "")
    )
)

fig.update_xaxes(
    title_text="",
    tickangle=-20
)

fig.update_yaxes(
    title_text="Median"
)

fig.update_layout(
    title={
        "x": 0.5,
        "xanchor": "center"
    },
    margin={
        "t": 100,
        "b": 100,
        "l": 80,
        "r": 40
    },
    showlegend=False
)

fig.show()

```

```python
import gradio as gr
import pandas as pd

# Feature order used during model training
imp_columns = ['HGB', 'MCH', 'RDW_CV', 'Gender_Male']

# Class mapping
class_mapping = {
    0: 'Anemic',
    1: 'Compensated Microcytosis',
    2: 'Non-Anemic'
}

def predict_anemia(HGB, MCH, RDW_CV, Gender_Male):

    # Prepare input in the same format as training data
    input_data = pd.DataFrame(
        [[HGB, MCH, RDW_CV, Gender_Male]],
        columns=imp_columns
    )

    # Predict class
    prediction = dt.predict(input_data)[0]

    # Convert predicted class into its label
    predicted_class = class_mapping[int(prediction)]

    return predicted_class


# Create Gradio interface
demo = gr.Interface(
    fn=predict_anemia,

    inputs=[
        gr.Number(label="Hemoglobin (HGB)", value=13.0),
        gr.Number(label="Mean Corpuscular Hemoglobin (MCH)", value=28.0),
        gr.Number(label="Red Cell Distribution Width (RDW_CV)", value=13.0),
        gr.Radio(
            choices=[
                ("Male", 1),
                ("Female", 0)
            ],
            label="Gender",
            value=1
        )
    ],

    outputs=gr.Textbox(
        label="Predicted Classification"
    ),

    title="Anemia Classification System",

    description="""
    Enter the patient's blood parameters to predict the anemia classification
    using a trained Decision Tree model.
    """,

    theme=gr.themes.Soft()
)

# Launch application
demo.launch(share=True)
```
